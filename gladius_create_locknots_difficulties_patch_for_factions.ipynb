```python
import re
```

```python
template_text = '''
  <entry name="Difficulty0ImperialFists" value="Neophyte"/>
	<entry name="Difficulty1ImperialFists" value="Battle-Brother"/>
	<entry name="Difficulty2ImperialFists" value="Sergeant"/>
	<entry name="Difficulty3ImperialFists" value="Captain-"/>
	<entry name="Difficulty4ImperialFists" value="Captain"/>
	<entry name="Difficulty5ImperialFists" value="Captain+"/>
	<entry name="Difficulty6ImperialFists" value="Chapter-Master-"/>
	<entry name="Difficulty7ImperialFists" value="Chapter-Master"/>
	<entry name="Difficulty8ImperialFists" value="Primarch-"/>
	<entry name="Difficulty9ImperialFists" value="Primarch"/>
	<entry name="Difficulty10ImperialFists" value="Emperor-"/>
	<entry name="Difficulty11ImperialFists" value="Emperor"/>
	<entry name="Difficulty12ImperialFists" value="Emperor+"/>
'''
template_faction_name = "ImperialFists"
template_dictionary_replacement = {
    0: "Neophyte",
    1: "Battle-Brother",
    2: "Sergeant",
    3: "Captain",
    4: "Chapter-Master",
    5: "Primarch",
    6: "Emperor",
}
```

```python
def create_dictionary_for_new_faction(string_from_world_parameters):
  new_difficulty_to_tier_dictionary = {}

  # Matches first instance of difficulty. Stops as soon as possible.
  match_object_first_line_for_name = re.search("\"Difficulty.*?\"", string_from_world_parameters)
  match_group_first_line_for_name = match_object_first_line_for_name.group()
  #print(match_group_first_line_for_name)
  # Now it will be "Difficulty0<faction_name>". Getting substring to only be faction name
  new_faction_name = match_group_first_line_for_name[12:-1]
  #print(new_faction_name)

  for i in range(0, 7):
    match_group_for_difficulty_line = re.search("\"Difficulty" + str(i) + ".*?\"/>", string_from_world_parameters).group()
    string_with_faction_and_value = match_group_for_difficulty_line[12:-3]
    # Now string is like <faction>" value="<target>
    # Find how long the first part is then cut it off
    non_target_span = re.search(".*value=\"", string_with_faction_and_value).span()
    non_target_span_end = non_target_span[1]
    target_only = string_with_faction_and_value[non_target_span_end:]
    #print(f"{i}: {target_only}")
    new_difficulty_to_tier_dictionary[i] = target_only

  return new_faction_name, new_difficulty_to_tier_dictionary

def create_text_for_new_faction(new_faction_name, new_difficulty_to_tier_dictionary):
  # Replaces faction name
  new_text = re.sub(template_faction_name, new_faction_name, template_text)
  for difficulty in new_difficulty_to_tier_dictionary:
    new_text = re.sub(template_dictionary_replacement[difficulty], new_difficulty_to_tier_dictionary[difficulty], new_text)
  return new_text
```

```python
original_faction_difficulties = input("Paste difficulty entries from <mod>/Data/Core/Languages/English/WorldParameters.xml for the relevant faction:\n")
print()

faction_name, difficulty_name_dictionary = create_dictionary_for_new_faction(original_faction_difficulties)
new_text = create_text_for_new_faction(faction_name, difficulty_name_dictionary)
print(new_text)
```
